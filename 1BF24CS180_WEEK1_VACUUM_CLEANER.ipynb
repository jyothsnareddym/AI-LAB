```python
## jyothsnareddy m
def vacuum_cleaner(location,A_status,B_status):
    while A_status == "Dirty" or B_status == "Dirty":
        if location == "A":
            if A_status == "Dirty":
                print("Location:", location)
                print("Action: Suck")
                A_status = "Clean"
            else:
                print("Location:", location)
                print("Action: Move Right")
                location = "B"
        else:
            if B_status == "Dirty":
                print("Location:", location)
                print("Action: Suck")
                B_status = "Clean"
            else:
                print("Location:", location)
                print("Action: Move Left")
                location = "A"
    print("Goal Achieved")
    print("A:", A_status)
    print("B:", B_status)
vacuum_cleaner("A", "Dirty", "Dirty")
```
